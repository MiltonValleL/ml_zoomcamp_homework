## Randomly Generated Arrays

```python
# 3 Random values from 0 to 1 (uniform distribution)
a = np.random.rand(3)
a
```

```python
# 5x2 Random values (uniform distribution)
b = np.random.rand(5, 2)
b
```

```python
# Now let's confirm the shape of our random values
print(f" - Shape of a:{a.shape}  \n - Shape of b:{b.shape}")
```

<br>

---

```python
# to get the same random values every time, we need to set a seed (uniform distribution)
np.random.seed(3)
np.random.rand(5, 2)
```

```python
# if we ask for another random values, with the same seed. we'll get the same matrix of random values
# (uniform distribution)
np.random.seed(3)
np.random.rand(5, 2)
```

```python
# Now we create random values from a "Standard Normal Distribution"
np.random.seed(3)
np.random.randn(5, 2)
```

```python
# Random values from a "Standard Normal Distribution"
np.random.seed(2)
np.random.randn(5, 2)
```

<br>

---

```python
# Random integer values from a "Discrete Uniform Distribution"
np.random.seed(2)
np.random.randint(low=0, high=100, size=(5, 2))
```

```python
# Random integer values from a "Discrete Uniform Distribution"
np.random.seed(2)
np.random.randint(low=0, high=100, size=(2, 5))
```

```python
# Random integer values from a "Discrete Uniform Distribution"
np.random.seed(2)
np.random.randint(low=0, high=100, size=(3, 5))
```

- Notice that If we use the same **random seed**, in this case `np.random.seed(2)`. but we change the size of the random integers, **we always get the same first ten values.**
